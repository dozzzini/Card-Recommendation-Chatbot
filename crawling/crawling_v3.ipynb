```python
from selenium import webdriver as wb
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

driver = wb.Chrome()
driver.get("https://www.card-gorilla.com/card?cate=CRD")

time.sleep(3)
body = driver.find_element(By.TAG_NAME, 'body')
# 페이지 스크롤
while True : 
    
    for _ in range(3): 
        body.send_keys(Keys.END)
        time.sleep(2) # 스크롤 후 로딩 대기 필수!

    try :
            btn = driver.find_element(By.XPATH, '//*[@id="q-app"]/section/div[1]/section/div/div[2]/div[2]/a[1]')
            btn.click()
            time.sleep(2)
    # 더보기 버튼이 없는 경우    
    except :
        print("더 이상 더보기 버튼이 없네요 ㅠ 그래도 코드는 이어서 동작합니다!")
        break

cards = driver.find_elements(By.CSS_SELECTOR, "#q-app > section > div.card > section > div > div.card_list > ul > li > div")
    #q-app > section > div.card > section > div > div.card_list > ul > li:nth-child(2) > div
    #q-app > section > div.card > section > div > div.card_list > ul > li:nth-child(3) > div
card_list = []
print(f"추출된 카드 개수: {len(cards)}")

for card in cards:
    try:
        name = card.find_element(By.CSS_SELECTOR, "div.card_data > div.name > p > span.card_name").text
        #q-app > section > div.card > section > div > div.card_list > ul > li:nth-child(2) > div > div.card_data > div.name > p > span.card_name
        # 카드사 정보도 함께 가져오기
        corp = card.find_element(By.CSS_SELECTOR, "div.card_data > div.name > p > span.card_corp").text
        #q-app > section > div.card > section > div > div.card_list > ul > li:nth-child(2) > div > div.card_data > div.name > p > span.card_corp
        # 혜택 정보
        benefits = card.find_element(By.CSS_SELECTOR, "div.card_data > div.sale").text
        
        #q-app > section > div.card > section > div > div.card_list > ul > li:nth-child(10) > div > div.card_data > div.sale
        # 연회비 정보
        membership = card.find_element(By.CSS_SELECTOR, "div.card_data > div.ex > p.in_for").text
        #q-app > section > div.card > section > div > div.card_list > ul > li:nth-child(10) > div > div.card_data > div.ex > p.in_for
        # 실적 정보
        performance = card.find_element(By.CSS_SELECTOR, "div.card_data > div.ex > p.l_mth").text
        #q-app > section > div.card > section > div > div.card_list > ul > li:nth-child(10) > div > div.card_data > div.ex > p.l_mth
            
        print(f"[{corp}] {name} | {benefits} | {membership} | {performance}")
    except Exception as e:
        print("데이터 추출 실패")

# 작업 완료 후 브라우저 종료
driver.quit()
```

```python
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
import time
import pandas as pd

def crawl_card_list_fast():
    # 1. 브라우저 실행
    options = uc.ChromeOptions()
    driver = uc.Chrome(options=options)
    
    try:
        url = "https://www.card-gorilla.com/card?cate=CRD"
        print(f"🔗 접속 중: {url}")
        driver.get(url)
        time.sleep(4) # 충분한 로딩 시간

        # 2. 스크롤 내리기 (모든 카드를 수집하려면 횟수를 늘리세요)
        print("📜 리스트를 아래로 내리고 있습니다...")
        for _ in range(3): 
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
            time.sleep(2)

        # 3. 카드 요소들 수집
        cards = driver.find_elements(By.CSS_SELECTOR, "li.card_con")
        print(f"📊 수집 대상: {len(cards)}개")

        card_data = []

        for card in cards:
            try:
                # 카드명, 회사명
                name = card.find_element(By.CLASS_NAME, "card_nm").text.strip()
                corp = card.find_element(By.CLASS_NAME, "corp_nm").text.strip()
                
                # 혜택 3개 (주요 키워드만)
                benefit_els = card.find_elements(By.CSS_SELECTOR, ".key_benefit span")
                benefits = [b.text.strip() for b in benefit_els if b.text.strip()][:3]
                
                # [핵심] 연회비 및 실적 정보 (리스트 하단 텍스트)
                # 보통 .card_info 또는 .bnf2 영역에 '해외 1.5만 / 실적 40만' 식으로 적혀 있습니다.
                # 카드고릴라 리스트에서는 'bnf2' 클래스 내부에 연회비와 실적이 위치함
                info_text = card.find_element(By.CLASS_NAME, "bnf2").text.replace('\n', ' ')
                
                # 예: "국내 10,000원 / 해외 12,000원 전월실적 40만원 이상"
                # 이를 조금 더 보기 좋게 분리 (문자열 파싱)
                annual_fee = "정보 없음"
                performance = "정보 없음"
                
                if "실적" in info_text:
                    parts = info_text.split("전월실적")
                    annual_fee = parts[0].strip()
                    performance = parts[1].strip()
                else:
                    annual_fee = info_text

                card_data.append({
                    "카드명": name,
                    "카드사": corp,
                    "혜택3개": ", ".join(benefits),
                    "연회비정보": annual_fee,
                    "실적조건": performance
                })
            except:
                continue

        return card_data

    finally:
        driver.quit()

# 실행
data = crawl_card_list_fast()
df = pd.DataFrame(data)

print("\n✅ 수집 완료!")
df
```
